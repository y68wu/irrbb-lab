# IRRBB case study
A synthetic Canadian banking book is exposed to the same rate shock in two ways: the economic value of today's contractual and behavioral cash flows moves immediately, while the change in net interest income accumulates through the year.

```python
from pathlib import Path
import json
import subprocess

import pandas as pd
from IPython.display import Image, display

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
reports = root / "reports"
summary = json.loads((reports / "summary.json").read_text())
display(Image(filename=str(root / "figures" / "irrbb-tradeoff.png")))
```

## Six-shock EVE screen
Compare the reference book with the sample swap notional at the same behavioral assumptions. Adverse EVE is measured against 15% of Tier 1 capital.

```python
unhedged = pd.read_csv(reports / "scenarios_unhedged.csv").set_index("label")
hedged = pd.read_csv(reports / "scenarios_hedged.csv").set_index("label")
eve = pd.DataFrame({"No swap (CAD M)": unhedged.deltaEve * 1000,
                    "Pay-fixed swap (CAD M)": hedged.deltaEve * 1000,
                    "Hedged loss / Tier 1 (%)": hedged.eveLossToTier1 * 100})
display(eve.round(2))
```
